```python
#RESULTADO
#pairs/
# ├── train/
# │    ├── pos_0_a.png, pos_0_b.png, ...
# │    ├── neg_0_a.png, neg_0_b.png, ...
# ├── val/
# │    ├── ...
# └── test/
#      ├── ...
#pairs_train.csv
#pairs_val.csv
#pairs_test.csv
import os
import random
import csv
from PIL import Image, ImageOps

# Ruta al dataset descargado con snapshot_download
# (ajústala según tu entorno)
splits = ["train", "val", "test"]

def augment(img):
    """Pequeñas transformaciones para simular 'mismo'."""
    if random.random() < 0.5:
        img = ImageOps.mirror(img)
    if random.random() < 0.5:
        img = ImageOps.autocontrast(img)
    return img

def generate_pairs(repo_path, split, n_pos=100, n_neg=100):
    img_dir = os.path.join(repo_path, split, "images")
    out_dir = os.path.join("/content", "pairs", split) #/content/sample_data
    os.makedirs(out_dir, exist_ok=True)

    csv_path = os.path.join("/content", f"pairs_{split}.csv")

    # Listar imágenes disponibles
    images = [os.path.join(img_dir, f) for f in os.listdir(img_dir)
              if f.endswith((".png", ".jpg", ".jpeg"))]

    if len(images) < 2:
        print(f"⚠️ Muy pocas imágenes en {split}, se omite.")
        return

    with open(csv_path, "w", newline="") as csvfile:
        writer = csv.writer(csvfile)
        writer.writerow(["img_a", "img_b", "label"])  # encabezado

        # ---- Pares positivos ----
        for i in range(n_pos):
            img_path = random.choice(images)
            img = Image.open(img_path).convert("RGB")

            img_a = img
            img_b = augment(img.copy())

            a_path = os.path.join(out_dir, f"pos_{i}_a.png")
            b_path = os.path.join(out_dir, f"pos_{i}_b.png")

            img_a.save(a_path)
            img_b.save(b_path)

            writer.writerow([a_path, b_path, 0])

        # ---- Pares negativos ----
        for i in range(n_neg):
            img1, img2 = random.sample(images, 2)
            img_a = Image.open(img1).convert("RGB")
            img_b = Image.open(img2).convert("RGB")

            a_path = os.path.join(out_dir, f"neg_{i}_a.png")
            b_path = os.path.join(out_dir, f"neg_{i}_b.png")

            img_a.save(a_path)
            img_b.save(b_path)

            writer.writerow([a_path, b_path, 1])

    print(f"✅ Generados {n_pos+n_neg} pares en {csv_path}")

# ---- Ejecutar para cada split ----
for split in splits:
    generate_pairs(repo_path, split, n_pos=100, n_neg=100)  # puedes cambiar cantidades de pares positivos y negativos
```

```python
import os
import random
from PIL import Image
import numpy as np
from tqdm import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
from sklearn.metrics import roc_auc_score, roc_curve

# -----------------------------
# 1) Dataset Siamese desde CSV
# -----------------------------
class SiameseUIDataset(Dataset):
    def __init__(self, pair_csv, transform=None):
        """
        pair_csv: ruta al CSV con columnas img_a,img_b,label
        transform: torchvision transforms
        """
        self.transform = transform
        self.pairs = []

        with open(pair_csv, "r") as f:
            next(f)  # saltar header
            for line in f:
                img_a, img_b, label = line.strip().split(",")
                self.pairs.append((img_a, img_b, int(label)))

    def __len__(self):
        return len(self.pairs)

    def __getitem__(self, idx):
        img_a_path, img_b_path, label = self.pairs[idx]
        img_a = Image.open(img_a_path).convert("RGB")
        img_b = Image.open(img_b_path).convert("RGB")

        if self.transform:
            img_a = self.transform(img_a)
            img_b = self.transform(img_b)

        return img_a, img_b, torch.tensor([label], dtype=torch.float32)

# -----------------------------
# 2) Transforms
# -----------------------------
transform = transforms.Compose([
    transforms.Resize((256,256)),
    transforms.RandomResizedCrop(224),
    transforms.RandomHorizontalFlip(p=0.1),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.1, hue=0.02),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225]),
])

# -----------------------------
# 3) Red Siamese
# -----------------------------
class SiameseNet(nn.Module):
    def __init__(self, embedding_size=256, pretrained=True):
        super().__init__()
        base = models.resnet18(pretrained=pretrained)
        modules = list(base.children())[:-1]  # quitar fc final
        self.backbone = nn.Sequential(*modules)
        self.head = nn.Sequential(
            nn.Flatten(),
            nn.Linear(base.fc.in_features, 512),
            nn.ReLU(),
            nn.Linear(512, embedding_size)
        )

    def forward_once(self, x):
        x = self.backbone(x)
        x = self.head(x)
        x = F.normalize(x, p=2, dim=1)
        return x

    def forward(self, a, b):
        ea = self.forward_once(a)
        eb = self.forward_once(b)
        return ea, eb

# -----------------------------
# 4) Contrastive Loss
# -----------------------------
class ContrastiveLoss(nn.Module):
    def __init__(self, margin=1.0):
        super().__init__()
        self.margin = margin

    def forward(self, e1, e2, label):
        d = F.pairwise_distance(e1, e2, p=2)
        loss_pos = (1 - label) * 0.5 * d.pow(2)
        loss_neg = label * 0.5 * F.relu(self.margin - d).pow(2)
        return (loss_pos + loss_neg).mean()

# -----------------------------
# 5) Entrenamiento
# -----------------------------
def train_siamese(pair_csv_train, pair_csv_val=None, epochs=10, batch_size=32, lr=1e-4, embedding_size=256):
    device = 'cuda' if torch.cuda.is_available() else 'cpu'

    # Dataset y loader
    train_ds = SiameseUIDataset(pair_csv_train, transform=transform)
    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True, num_workers=4, pin_memory=True)

    if pair_csv_val:
        val_ds = SiameseUIDataset(pair_csv_val, transform=transform)
        val_loader = DataLoader(val_ds, batch_size=batch_size, shuffle=False, num_workers=4, pin_memory=True)
    else:
        val_loader = None

    # Modelo, loss, optim
    model = SiameseNet(embedding_size=embedding_size, pretrained=True).to(device)
    crit = ContrastiveLoss(margin=1.0)
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=7, gamma=0.5)

    # Entrenamiento
    for epoch in range(epochs):
        model.train()
        running_loss = 0.0
        for a,b,label in tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs}"):
            a, b, label = a.to(device), b.to(device), label.to(device)
            optimizer.zero_grad()
            ea, eb = model(a,b)
            loss = crit(ea, eb, label)
            loss.backward()
            optimizer.step()
            running_loss += loss.item() * a.size(0)

        scheduler.step()
        avg_loss = running_loss / len(train_loader.dataset)
        print(f"Epoch {epoch+1} train loss: {avg_loss:.4f}")

        if val_loader:
            auc, thr = evaluate_embeddings(model, val_loader, device)
            print(f"Validation AUC: {auc:.4f}, suggested threshold: {thr:.3f}")

    return model

# -----------------------------
# 6) Evaluación
# -----------------------------
def evaluate_embeddings(model, loader, device='cpu'):
    model.eval()
    dists, labels = [], []
    with torch.no_grad():
        for a,b,label in loader:
            a, b = a.to(device), b.to(device)
            ea, eb = model(a,b)
            dist = F.pairwise_distance(ea, eb).cpu().numpy()
            dists.extend(dist.tolist())
            labels.extend(label.squeeze(1).numpy().tolist())

    auc = roc_auc_score(labels, dists)
    fpr, tpr, thr = roc_curve(labels, dists)
    youden = tpr - fpr
    best_idx = np.argmax(youden)
    best_thr = thr[best_idx]
    return auc, best_thr

# -----------------------------
# 7) Inferencia
# -----------------------------
def detect_change(model, img_a_path, img_b_path, threshold, device='cpu'):
    t = transforms.Compose([
        transforms.Resize((224,224)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225])
    ])
    img_a = Image.open(img_a_path).convert("RGB")
    img_b = Image.open(img_b_path).convert("RGB")
    a = t(img_a).unsqueeze(0).to(device)
    b = t(img_b).unsqueeze(0).to(device)
    model.eval()
    with torch.no_grad():
        ea, eb = model(a,b)
        d = F.pairwise_distance(ea, eb).item()
    return d, d > threshold
```

```python
# -----------------------------
# 8) Ejemplo de uso
# -----------------------------
if __name__ == "__main__":
    repo_path = "/content"  # cambia por tu path local
    pair_csv_train = os.path.join(repo_path, "pairs_train.csv")
    pair_csv_val   = os.path.join(repo_path, "pairs_val.csv")
    pair_csv_test  = os.path.join(repo_path, "pairs_test.csv")

    model = train_siamese(pair_csv_train, pair_csv_val, epochs=15, batch_size=32)

    # Evaluar en test
    test_ds = SiameseUIDataset(pair_csv_test, transform=transform)
    test_loader = DataLoader(test_ds, batch_size=64, shuffle=False)
    auc, thr = evaluate_embeddings(model, test_loader)
    print(f"Test AUC: {auc:.4f}, threshold sugerido: {thr:.3f}")
```
